```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('AcademicPerformance.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['PhD'] = df['PhD'].fillna(df['PhD'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.boxplot(df['Marks'])
plt.title("Boxplot of Marks")
plt.show()
```

```python
Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Marks'] < lower) | (df['Marks'] > upper)]

outliers
```

```python
df = df[(df['Marks'] >= lower) & (df['Marks'] <= upper)]
df
```

```python
plt.boxplot(df['Age'])
plt.title("Boxplot of Age")
plt.show()
```

```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Age'] < lower) | (df['Age'] > upper)]
outliers
```
