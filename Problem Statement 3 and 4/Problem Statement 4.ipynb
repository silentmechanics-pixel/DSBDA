```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('AcademicPerformance.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['PhD'] = df['PhD'].fillna(df['PhD'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df['Marks'].skew()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['Marks'])
plt.title("Marks before transformation")
plt.show()
```

```python
df['Marks_transform'] = np.sqrt(df['Marks'])
```

```python
plt.hist(df['Marks_transform'])
plt.title("Marks after transformation")
plt.show()
```
