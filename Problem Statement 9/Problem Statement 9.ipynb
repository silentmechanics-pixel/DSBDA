## Tokenization

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk import word_tokenize, sent_tokenize

# Subject-based sentence
sent = "Data science is one of the most important fields in computer engineering. Machine learning plays a key role in data science."

print("Word Tokenization:\n", word_tokenize(sent))
print("\nSentence Tokenization:\n", sent_tokenize(sent))
```

## Stop Words Removal

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
stop_words = stopwords.words('english')
print("\nStopwords:\n", stop_words)
```

## Cleaning Text (removing stopwords)

```python
token = word_tokenize(sent)
cleaned_token = []

for word in token:
    if word.lower() not in stop_words:
        cleaned_token.append(word)

print("\nOriginal Tokens:\n", token)
print("\nAfter Stopword Removal:\n", cleaned_token)
```

```python
#Lowercase and punctuation removal
words = [word.lower() for word in cleaned_token if word.isalpha()]
print("\nClean Words:\n", words)
```

## Stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
port_stemmer_output = [stemmer.stem(word) for word in words]

print("\nStemming Output:\n", port_stemmer_output)
```

## Lemmatization

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
lemmatizer_output = [lemmatizer.lemmatize(word) for word in words]

print("\nLemmatization Output:\n", lemmatizer_output)
```

## POS Tagging

```python
from nltk import pos_tag
nltk.download('averaged_perceptron_tagger_eng')
```

```python
tagged = pos_tag(cleaned_token)

print("\nPOS Tagging:\n", tagged)
```

## TF-IDF Vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = [
    "Data science is an important field",
    "Machine learning is a part of data science",
    "Deep learning is used in machine learning",
    "Artificial intelligence is related to data science"
]
```

```python
vectorizer = TfidfVectorizer(analyzer='word', norm=None, use_idf=True, smooth_idf=True)

# Vocabulary
vectorizer.fit(docs)
print("\nVocabulary:\n", vectorizer.vocabulary_)
```

```python
# TF-IDF Matrix
tfidfMat = vectorizer.fit_transform(docs)

print("\nTF-IDF Matrix:\n", tfidfMat)
```
