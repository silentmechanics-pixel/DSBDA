```python
import csv
```

```python
temp_sum = 0
dew_sum = 0
wind_sum = 0
count = 0

with open("weather.txt", "r") as file:
    reader = csv.reader(file)
    next(reader)

    for row in reader:
        try:
            temp = float(row[1])
            dew = float(row[2])
            wind = float(row[3])

            temp_sum += temp
            dew_sum += dew
            wind_sum += wind
            count += 1

        except:
            continue

print("Average Temperature: ",temp_sum/count)
print("Average Wind: ",wind_sum/count)
print("Average dew: ",dew_sum/count)
```
